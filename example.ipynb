```python
from pyomo.environ import *
```

```python
model = ConcreteModel()
```

```python
model.cake = Var(within=NonNegativeReals)
model.cookies = Var(within=NonNegativeReals)
```

```python
model.maximizeProfit = Objective(expr = 40*model.cake + 25*model.cookies, sense=maximize)
```

```python
model.FlourConstraint = Constraint(expr = 3*model.cake + 2*model.cookies <=180)
model.TimeConstraint = Constraint(expr = 2*model.cake + model.cookies <=100)
model.LabourConstraint = Constraint(expr = model.cake + 2*model.cookies <=120)
```

```python
optimizer = SolverFactory("appsi_highs")
optimizer.solve(model)
model.display()
```
